## Seaborn

#### Its a powerful visualization library built on top of Matplotlib
#### Seaborn complements Matplotlib. It enhances visuals

```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()
```

```python
# Load the titnic dataset

data = sns.load_dataset("titanic")
```

```python
data.head()
```

#### Histogram

```python
plt.figure(figsize=(10, 6))
sns.histplot(data["age"], kde=True, color="Brown") # kde is the distribution line
plt.xlabel("Age of the Passengers")
plt.ylabel("Frequency")
# plt.grid(True)
plt.show()
```

```python
sns.set(style="darkgrid", palette="viridis", context="paper", font="Arial")

plt.figure(figsize=(10, 6))
sns.histplot(data["age"], kde=True, color="Brown") # kde is the distribution line
plt.xlabel("Age of the Passengers")
plt.ylabel("Frequency")
# plt.grid(True)
plt.show()
```

#### CountPlot

```python
data.head()
```

```python
data["sex"].value_counts()
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(data["sex"], color="brown")
plt.show()
```

```python
# vertical
sns.set(style="darkgrid", palette="viridis", context="paper", font="Arial")
plt.figure(figsize=(8, 6))
sns.countplot(x=data["sex"], color="brown")
plt.title("Distribution of Gender of the Titlanic Passengers")
plt.xlabel("Sex")
plt.ylabel("Frequnecy")
plt.show()
```

#### Bar Plot

```python
data.head()
```

```python
sns.set(style="darkgrid", palette="viridis", context="paper", font="Arial", font_scale=1.2)

plt.figure(figsize=(8, 6))
sns.barplot(x=data["sex"], y=data["survived"], color="brown", estimator="mean", errorbar=None, width=0.5)
plt.title("Average survival rate by Gender on the Titlanic")
plt.xlabel("Sex")
plt.ylabel("Survival Rate")
plt.show()
```

```python
sns.set(style="darkgrid", palette="viridis", context="paper", font="Arial", font_scale=1.2)

plt.figure(figsize=(8, 6))
sns.barplot(x=data["embark_town"], y=data["fare"], color="brown", width=0.5) # errorbar=None: removes the black lines
plt.title("Fares by town passengers embarked from")
plt.xlabel("Embarked From")
plt.ylabel("Fares")
plt.show()
```

```python
sns.set(style="darkgrid", palette="viridis", context="paper", font="Arial", font_scale=1.2)

plt.figure(figsize=(8, 6))
sns.barplot(x=data["class"], y=data["survived"], color="darkgreen", errorbar=None, width=0.5) # errorbar=None: removes the black lines
plt.title("Survival Rates by Class on the Titanic")
plt.xlabel("Class")
plt.ylabel("Survival Rate")
plt.show()
```

#### Grouped Bar

```python
sns.set(style="darkgrid", palette="viridis", context="paper", font="Arial", font_scale=1.2)

plt.figure(figsize=(8, 6))
sns.barplot(x=data["class"], y=data["survived"], hue=data["sex"], palette="Set1", errorbar=None, width=0.8)
plt.title("Survival Rates by Class on the Titanic")
plt.xlabel("Class")
plt.ylabel("Survival Rate")
plt.show()
```

#### Heatmap

```python
data.head()
```

```python
numerical_data = data.select_dtypes(include="number")
```

```python
numerical_data.head()
```

```python
# correlation between two variables
correlation_map = numerical_data.corr()
correlation_map

# strongest correction: 1.0. Both variables are increasing. same direction
# Inverse correlation: -1.0. As one increases, the other decreases. Inverse direction
# zero: no correlation at all
```

```python
plt.figure(figsize=(12, 6))
sns.heatmap(correlation_map, annot=True, fmt=".2f", linewidths=0.5)
plt.title("Correlation of all variables")
plt.show()
```

#### ScatterPlot

##### Correlation between two variables as well

```python
plt.figure(figsize=(12, 6))
sns.scatterplot(x=data["age"], y=data["fare"])
plt.title("Correlation of age vs fare variables")
plt.show()

# No correlation. Age and fare is 0.10 on the heatmap. close to 0, no correlation as shown below. Its not stable
# They go the opposite direction in an unstable manner
```

#### Regression Plot
##### Similar to scatter plot but with a line to demonstrate the correlation

```python
plt.figure(figsize=(12, 6))
sns.regplot(x=data["age"], y=data["fare"])
plt.title("Correlation of age vs fare variables")
plt.show()
```

####  Violin Plot

#### Box Plot

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x=data["sex"], y=data["age"])
plt.title("Correlation of age vs fare variables")
plt.show()

# shows alot of the passengers are between 20 and 40
```

#### Swarm Plot

```python
plt.figure(figsize=(12, 6))
sns.swarmplot(x=data["sex"], y=data["age"])
plt.title("Correlation of age vs fare variables")
plt.show()

# shows alot of the passengers are between 20 and 40
```

## Visualization with built-in Pandas (Seaborn)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()
```

```python
#  Load the tips data
data = sns.load_dataset('tips')
data.head()
```

#### Line Plot

```python
data["total_bill"].plot(kind="line", figsize=(10, 6), color="blue", title="Trend of Total Bill Over all Entries")
plt.xlabel("Index")
plt.ylabel("Total Bill")
plt.show()
```

```python
data["tip"].plot(kind="line", figsize=(10, 6), color="blue", title="Trend of Tip Over all Entries")
plt.xlabel("Index")
plt.ylabel("Amount Tip")
plt.show()
```

```python
data.head(10)
```

#### Bar Graph

```python
# Calculated the average amount tipped per day
avg_tip_per_day = data.groupby("day")["tip"].mean().round(2)
avg_tip_per_day
```

```python
avg_tip_per_day.plot(kind="bar", figsize=(10, 6), color="darkgreen", title="Average amount tipped per day")
```

```python
# horizontal version
avg_tip_per_day.plot(kind="barh", figsize=(10, 6), color="darkgreen", title="Average amount tipped per day")
```

```python
data.head()
```

#### Group Bar chart

```python
# three different variables
# what gender is paying the most or least per day

# calculate the ave amount tipped by day by gender

avg_tip_per_day_sex = data.groupby(["day","sex"])["tip"].mean().unstack() # stacks them so same day dont appear multiple times
avg_tip_per_day_sex
```

```python
avg_tip_per_day_sex.plot(kind="bar", color=["darkgreen", "darkblue"], title="Average Amount Tipped per Day by Gender")
```

```python
data.head(5)
```

```python
# Total amount tipped per day by smoking status
tip_per_day_smoker = data.groupby(["day","smoker"])["tip"].sum().unstack()
tip_per_day_smoker
```

```python
tip_per_day_smoker.plot(kind="bar", stacked=True, color=["darkgreen", "darkblue"], title="Total Amount Tipped per Day by Smoking status")
# stacked=True stacked the bins on each other
```

#### Histogram

```python
# Distribution of the total bill
data.head(2)
```

```python
data["total_bill"].plot(kind="hist", bins=8, color="darkgreen", title="Distribution of the tota bill")
```

#### Box Plot

```python
# Amount tipped per day
data.boxplot(column="tip", by="day", grid=False, color="darkblue", figsize=(8,5))
```

#### Scatter Plot

```python
# Total Bill Vs Amount Tipped
data.plot(kind="scatter", x="total_bill", y="tip", color="violet", alpha=0.8, title="Total Bill Vs Amount Tipped",figsize=(8,6))
```

#### Pie Chart

```python
# Total Amount tipped per day
total_daily_tips = data.groupby("day")["tip"].sum()
total_daily_tips
```

```python
total_daily_tips.plot(kind="pie", autopct="%1.1f%%", figsize=(8,8), colormap="Pastel1")
```

#### Area Plot

```python
# Average Amount Tipped per Day
avg_daily_tips = data.groupby("day")["tip"].mean()
avg_daily_tips
```

```python
avg_daily_tips.plot(kind="area", color="darkgreen", grid=False,alpha=0.6, title="Average Amount Tipped per Day", ylabel="Tips")
```
