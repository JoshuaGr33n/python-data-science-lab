```python
import pandas as pd

# Sample DataFrame with sales, expenses, and profit data for different regions over several months
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales_North': [200, 220, 250, 270, 300, 310],
    'Sales_South': [180, 190, 230, 240, 260, 275],
    'Expenses_North': [150, 160, 180, 200, 220, 230],
    'Expenses_South': [130, 140, 160, 170, 185, 190],
    'Profit_North': [50, 60, 70, 70, 80, 80],
    'Profit_South': [50, 50, 70, 70, 75, 85]
}

# Convert to DataFrame
df = pd.DataFrame(data)

df
```

```python
# 2x1 subplot

fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(10, 8))
ax[0].plot(df["Month"], df["Sales_North"], color="blue", marker="o") # ax[0] represents the nrow(first row of the 2 rows)
ax[0].set_title("Monthly Sales in the Northern Region")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Sales")

ax[1].plot(df["Month"], df["Sales_South"], color="red", marker="s") # ax[1] represents the nrow(second row of the 2 rows)
ax[1].set_title("Monthly Sales in the Southern Region")
ax[1].set_xlabel("Month")
ax[1].set_ylabel("Sales")


plt.tight_layout()
plt.show()
```

```python
df.head()
```

```python
# 2x1 subplot

fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(10, 8))
ax[0].plot(df["Month"], df["Expenses_North"], color="blue", marker="o") # ax[0] represents the nrow(first row of the 2 rows)
ax[0].set_title("Monthly Expenses in the Northern Region")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Expenses")

ax[1].plot(df["Month"], df["Expenses_South"], color="red", marker="s") # ax[1] represents the nrow(second row of the 2 rows)
ax[1].set_title("Monthly Expenses in the Southern Region")
ax[1].set_xlabel("Month")
ax[1].set_ylabel("Expenses")


plt.tight_layout()
plt.show()
```

```python
# Join two charts in one subplot
fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(10, 8))
ax[0].plot(df["Month"], df["Sales_North"], color="blue", marker="o", label="Northern Sales") # ax[0] represents the nrow(first row of the 2 rows)
ax[0].plot(df["Month"], df["Sales_South"], color="orange", marker="s", label="Southern Sales")
ax[0].set_title("Monthly Sales by Region")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Sales")
ax[0].legend()

ax[1].plot(df["Month"], df["Expenses_North"], color="red", marker="o", label="Northern Expenses") # ax[1] represents the nrow(second row of the 2 rows)
ax[1].plot(df["Month"], df["Expenses_South"], color="green", marker="s", label="Southern Expenses")
ax[1].set_title("Monthly Expenses by Region")
ax[1].set_xlabel("Month")
ax[1].set_ylabel("Expenses")
ax[1].legend()


plt.tight_layout()
plt.show()
```

```python
# Join two charts in one subplot. Side by Side version
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(15, 5))
ax[0].plot(df["Month"], df["Sales_North"], color="blue", marker="o", label="Northern Sales") # ax[0] represents the nrow(first row of the 2 rows)
ax[0].plot(df["Month"], df["Sales_South"], color="orange", marker="s", label="Southern Sales")
ax[0].set_title("Monthly Sales by Region")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Sales")
ax[0].legend()

ax[1].plot(df["Month"], df["Expenses_North"], color="red", marker="o", label="Northern Expenses") # ax[1] represents the nrow(second row of the 2 rows)
ax[1].plot(df["Month"], df["Expenses_South"], color="green", marker="s", label="Southern Expenses")
ax[1].set_title("Monthly Expenses by Region")
ax[1].set_xlabel("Month")
ax[1].set_ylabel("Expenses")
ax[1].legend()


plt.tight_layout()
plt.show()
```

```python
# Join two charts in one subplot. Side by Side version
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(15, 5), sharey=True) # sharex for X axis

# Northern Region
ax[0].plot(df["Month"], df["Sales_North"], color="blue", marker="o", label="Northern Sales")
ax[0].plot(df["Month"], df["Expenses_North"], color="orange", marker="o", label="Northern Expenses")
ax[0].plot(df["Month"], df["Profit_North"], color="green", marker="o", label="Northern Profits")

ax[0].set_title("Monthly Analysis in the Northern Region")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Amount")
ax[0].legend()

# Southern Region
ax[1].plot(df["Month"], df["Sales_South"], color="blue", marker="s", label="Southern Sales")
ax[1].plot(df["Month"], df["Expenses_South"], color="orange", marker="s", label="Southern Expenses")
ax[1].plot(df["Month"], df["Profit_South"], color="green", marker="s", label="Southern Profits")

ax[1].set_title("Monthly Analysis in the Southern Region")
ax[1].set_xlabel("Month")
ax[1].set_ylabel("Amount")
ax[1].legend()

plt.tight_layout()
plt.subplots_adjust(wspace=0.3)
plt.show()
```

#### 2D Subplotting
Using 2x2 Subplot layout for Broader View. A complete financial picture

```python
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(12, 10))
```

```python
ax
```

```python
# Using flatten() to convert from 2D to 1D
fig, ax2d = plt.subplots(nrows=2, ncols=2, figsize=(12, 10))
ax1d = ax2d.flatten()

ax1d # flattened to 1d
```

```python
df.head()
```

##### Without flattening: ax[0,0] in multiple dimensions. single dimensions may not need flattening.
##### But with flattening only ax[0] is needed

```python
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(12, 10))
ax = ax.flatten()

ax[0].plot(df["Month"], df["Sales_North"], color="blue", marker="o")
ax[0].set_title("Northern Region Sales")
ax[0].set_xlabel("Months")
ax[0].set_ylabel("Sales")

ax[1].plot(df["Month"], df["Sales_South"], color="orange", marker="o")
ax[1].set_title("Southern Region Sales")
ax[1].set_xlabel("Months")
ax[1].set_ylabel("Sales")

ax[2].plot(df["Month"], df["Expenses_North"], color="red", marker="o")
ax[2].set_title("Northern Region Expenses")
ax[2].set_xlabel("Months")
ax[2].set_ylabel("Sales")

ax[3].plot(df["Month"], df["Expenses_South"], color="green", marker="o")
ax[3].set_title("Southern Region Expenses")
ax[3].set_xlabel("Months")
ax[3].set_ylabel("Sales")

plt.tight_layout()
plt.show()
```
