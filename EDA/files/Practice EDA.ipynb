```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('Student_performance_data.csv')
```

```python
data.head()
```

```python
data.drop('StudentID', axis=1, inplace=True)
```

```python
numerical_data = data.select_dtypes(include='number')
```

```python
numerical_data.head()
```

```python
fig, axes = plt.subplots(nrows=5, ncols=2, figsize=(20, 25))
ax = axes.flatten()

for index, column_name in enumerate(numerical_data):
    sns.histplot(data[column_name], ax=ax[index], kde=True, color="Orange")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(nrows=5, ncols=2, figsize=(20, 25))
ax = axes.flatten()

for index, column_name in enumerate(numerical_data):
    sns.boxplot(data[column_name], ax=ax[index], color="Orange")

plt.tight_layout()
plt.show()
```

```python
categorical_data = data.select_dtypes(include='object')
```

```python
categorical_data.head()
```

```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 6))
ax = axes.flatten()

for index, column_name in enumerate(categorical_data):
    sns.countplot(x=data[column_name], ax=ax[index], color="Orange")

plt.tight_layout()
plt.show()
```

### Bivariate Analysis
Bivariate analysis examines the relationship between two variables, typically with one independent variable and one dependent variable.

Numerical vs Numerical:
- We can use scatter plots and compute the correlation.

Numerical vs Categorical:
- We can use bar plot, box plots, violin plots, or swarm plots.
  
Categorical vs Categorical:
- We can use stacked bar plots or grouped bar plots.
