### Data Cleaning

```python
# Convert date columns from object (string) to datetime
df["Trip_Date"] = pd.to_datetime(df["Trip_Date"], errors="coerce") #errors="coerce" converts invalid dates into NaT (Not a Time), allowing us to identify problematic records.
df["Departure_Time"] = pd.to_datetime(df["Departure_Time"], errors="coerce")

# Verify the updated data types
df[["Trip_Date", "Departure_Time"]].dtypes
```

```python
# Check for invalid or missing dates after datetime conversion
df[["Trip_Date", "Departure_Time"]].isna().sum()
```

```python
# Compare Trip_Date with the date extracted from Departure_Time
df["Date_Match"] = (
    df["Trip_Date"].dt.normalize() == df["Departure_Time"].dt.normalize()
)

# Count matching and mismatching dates
df["Date_Match"].value_counts()
```

```python
# Inspect records where Trip_Date and Departure_Time do not match
df.loc[
    ~df["Date_Match"],
    ["Trip_ID", "Trip_Date", "Departure_Time", "Day_of_Week"]
].head(10)
```

```python
# Calculate the difference in days between Trip_Date and Departure_Time
df["Date_Difference"] = (
    df["Trip_Date"].dt.normalize() -
    df["Departure_Time"].dt.normalize()
).dt.days

# Examine how frequently each difference occurs
df["Date_Difference"].value_counts().sort_index()
```

The date validation revealed significant inconsistencies between Trip_Date and Departure_Time. Only 24 out of 1,000 records have matching dates, while the remaining 976 show differences ranging from 1 to 41 days. The differences follow a systematic pattern, suggesting inconsistent date generation or recording. Further investigation is required before selecting a reliable date field for time-based analysis.

```python
# Calculate net trip flow by station
departures = df["Departure_Station"].value_counts()
arrivals = df["Arrival_Station"].value_counts()

station_flow = pd.DataFrame({
    "Departures": departures,
    "Arrivals": arrivals
}).fillna(0)

station_flow["Net_Flow"] = (
    station_flow["Arrivals"] - station_flow["Departures"]
)

# Visualize net trip flow by station
fig, ax = plt.subplots(figsize=(12, 5))

sns.barplot(
    data=station_flow.reset_index(),
    x="Net_Flow",
    y="index",
    color="steelblue",
    ax=ax
)

ax.axvline(0, color="red", linestyle="--")

ax.set_title("Net Trip Flow by Station")
ax.set_xlabel("Net Trips (Arrivals - Departures)")
ax.set_ylabel("Station")

plt.show()
```

```python
# Examine correlations between the main numerical variables
# Use originally recorded values to avoid imputation effects

numeric_cols = [
    "Passenger_Count",
    "Fare_Amount",
    "Trip_Duration_Minutes"
]

# Exclude rows containing imputed values
original_data = df.loc[
    ~df["Passenger_Count_Imputed"] &
    ~df["Fare_Amount_Imputed"] &
    ~df["Trip_Duration_Imputed"]
]

# Calculate Pearson correlation matrix
corr_matrix = original_data[numeric_cols].corr()

# Visualize correlation matrix
fig, ax = plt.subplots(figsize=(8, 6))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".3f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    square=True,
    ax=ax
)

ax.set_title("Correlation Between Passenger Count, Fare Revenue and Trip Duration")

plt.tight_layout()
plt.show()
```
