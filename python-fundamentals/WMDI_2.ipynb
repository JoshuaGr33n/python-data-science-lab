### Bring all Data Types Together

```python
from datetime import date
```

```python
# Buiding a simple program for 10Alytics

# String
print("Welcome to 10Alytics") # By default the input returns a STRING
name = input("What is your first name: ")
eductation = input("What did you study in the university/college: ")

#Integer
yob = int(input("Please input your year of birth: "))
year = date.today().year
age = year - yob

# Float
# Rate your skillset
python = float(input("What is your proficiency in Python (0-10): ")) # float is better when integer or float is needed
excel = float(input("What is your proficiency in Excel (0-10): "))
tableau = float(input("What is your proficiency in Tableau (0-10): "))
sql = float(input("What is your proficiency in SQL (0-10): "))

# List
course = input("Provide a list of courses you are intereseted in. (Separate each course with a comma): ")
course_list = course.split(", ") # the comma acknoleges the commas and the space acknoleges the space. this ensures it groups words after every comma

# Set
skills = input("Enter your skillset (Separated by a comma (e.g Python, Excel, ML, Tableau etc)): ")
skill_list = skills.split(", ") # Convert it to list first before converting to set
skills_set = set(skill_list)

# Dictionary
full_name = input("What is your full_name: ")
location = input("What is your current location: ")
phone = input("Please enter your phone number: ")
email = input("What is your email address: ")

full_details = {"Full Name": full_name, "Location": location, "Phone Number": phone, "Email Address": email}

# Boolean
full_time = input("Do you want to learn full time (yes/no): ")
answer = full_time.strip().lower()
full_time_bool = answer == "yes"


print("------------------------------------------------------------------------")
print(f"Hello {name}, welcome to 10Alytics. You studied {eductation} at the university")
print(f"Your year of birth is {yob} which makes you {age} years old")
print(f"Python: {python}, Excel: {excel}, Tableau: {tableau}, SQL: {sql}")
print(f"Your course preferences are {course_list}")
print(f"Your skillset are {skills_set}")
print(f"Your full details are: {full_details}")
print(f"Full time learning: {full_time_bool}")
```
