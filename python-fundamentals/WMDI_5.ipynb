## FOR Loop Extra

```python
# Password verification
correct_password = "password123"
max_attempt = 5

for attempt in range(max_attempt):
    password_try = input(f"This is {attempt + 1} of {max_attempt}")

    if password_try == correct_password:
        print("Access Granted")
        break # correct password. end the attempts
    else:
        print("Incorrect Password")

        if attempt < max_attempt - 1:
            print("Please Try Again \n")
            continue # wrong password, continue to the next attempt
        else: 
            print("You have used all available attempts")
```

### WHILE Loop Extra

```python
# Guess Game
import random

lotto_number = random.randint(1,100)
print("We have chosen a random number between 1 and 100. Try to guess it")

while True:
    guess = input("Enter your guess(number) or type 'yes' to quit")

    # check if player wants to quit
    if guess.lower() == 'yes':
        print(f"The number to guess was {lotto_number}. Better luck next time")
        break

    # check if the input is a number
    if guess.isdigit():
        guess = int(guess)
    else:
        print("Please enter a valid number. Try again: ")
        continue

    # confirm guess number
    if guess < lotto_number:
        print("Your guess is lower than the lotto number")
    elif guess > lotto_number:
         print("Your guess is higher than the lotto number")
    else: 
         print("Congratulations! You guessed the right number! You are the winner")
         break
```
