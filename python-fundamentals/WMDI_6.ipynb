```python
def guess_game():
    
    # Guess Game
    import random

    lotto_number = random.randint(1,100)
    print("We have chosen a random number between 1 and 100. Try to guess it")

    while True:
        guess = input("Enter your guess(number) or type 'yes' to quit")

        # check if player wants to quit
        if guess.lower() == 'yes':
            print(f"The number to guess was {lotto_number}. Better luck next time")
            break

        # check if the input is a number
        if guess.isdigit():
            guess = int(guess)
        else:
            print("Please enter a valid number. Try again: ")
            continue

        # confirm guess number
        if guess < lotto_number:
            print("Your guess is lower than the lotto number")
        elif guess > lotto_number:
            print("Your guess is higher than the lotto number")
        else: 
            print("Congratulations! You guessed the right number! You are the winner")
            break
```

```python
guess_game()
```

```python
# Loan verification application

def comprehensive_loan_approval():

    # Criteria for loan approval
    min_credit_score = 700
    min_income = 30000 # per annum
    max_DTI = 0.4 # Debt to income ratio = 40%
    min_down_payment = 0.2 # 20%
    employment_status = ['employed', 'self-employed']
    min_length_of_employment = 2

    # Get user input
    credit_score = int(input("What is your credit score: "))
    income = float(input("What is your current income: "))
    loan_amount = float(input("Enter the loan amount you wish to borrow: "))
    down_payment = float(input("What is your down payment: "))
    employment = input("Enter your employment status (employed, self-employed, not employed): ")
    length_of_employment = int(input("How long have you been employed in your current job: "))

    # Check user input against the minimum criteria
    if credit_score < min_credit_score and income < min_income:
        status = "Loan denied: Credit score or Minimum Income is too low"
    elif employment.lower() not in employment_status:
        status = "Loan denied: Employment status not eligible"
    elif loan_amount > income * max_DTI:
            status = "Loan denied: Debt to income not eligible"
    elif down_payment > loan_amount * min_down_payment:
                status = "Loan denied: Down payment too low"
    elif length_of_employment > min_length_of_employment:
                    status = "Loan denied: length of current employment is too low"
    else:
        status = "Loan Approved"

    return status
```

```python
comprehensive_loan_approval()
```
