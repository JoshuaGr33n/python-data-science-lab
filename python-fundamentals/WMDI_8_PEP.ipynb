#### Managing Indentations

```python
# Example 1
def calculate_area(shape, dimension):
    if shape == 'circle': # Indented
        radius = dimension[0] # Indented
        return 3.14 * radius**2
    elif shape == 'square':
        side = dimension[0]
        return side * side
    else:
        return "Shape not recognised"
```

```python
calculate_area('circle',[2] )
```

```python
# Example 2
def calculate_volume(length, width, height):
    if length > 0 and width > 0 and height > 0:
        volume = length * width * height
        return volume
    else:
        return "Invalid dimension"
```

```python
calculate_volume(1,2,3)
```

```python
# Example 3
def check_eligibilty(age, citizenship):
    if age >= 18 and citizenship == 'Yes':
        return 'Eligible to vote'
    else:
        return 'Not eligible to vote'
```

```python
 check_eligibilty(18, "Yes")
```

```python
# Example 4
def list_even_numbers(numbers):
    for number in numbers:
        if number % 2 == 0:
            print(number)
        else:
            continue
```

```python
numbers = [1,22,45,98,120,250]
list_even_numbers(numbers)
```

#### Utilizing comments

```python
# Block comment
def list_even_numbers(numbers): # Inline comment
    for number in numbers:
        if number % 2 == 0:
            print(number)# Inline comment
        else:
            continue # Inline comment
```

```python
# Create function to check for prime numbers
def is_prime(number):
    if number > 1:
        for i in range(2, number):
            if (number % i) == 0:
                print(f"{number} is not a prime number")
                break
        else:
            print(f"{number} is a prime number")
    else:
        print(f"{number} is not a prime number")

```

```python
is_prime(21)
```

#### Documenting with Doc Strings

```python
# Documents a function, class, or module
# Comments are for one line. Docstring allow for more detailed comments and meta data
```

```python
# Single line docstring

def greet(name):
    """Greet is a function that returns an hello to an argument 'name'"""
    print(f"Hello, {name}")
```

```python
greet("John Doe")
```

```python
# Multi line docstring
def divide(x,y):
    """Divides the first number by the second and returns the result.
    
    Parameters
    ------------
    x:float
        The numerator
    y:float
        The denominator, should not be zero
    
    Returns
    ------------
    result:float
        The division of x by y
    """

    result = x/y
    return result
```

```python
divide(6,9)
```
