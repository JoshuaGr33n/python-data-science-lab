#### Filter, Map, and Lambda Function

```python
# Filter Function
# use filter() when you need to filter

# Extract the even numbers from the list
number = list(range(21))

def even_number(x):
    return x % 2 == 0

even_num = filter(even_number, number)

#convert into a list
list(even_num)
```

```python
words = ["Hi", "Hello", "World", "level", "radar"]

# Create a function to confirm if a word is a palindrome(spelt backwards returns same word)
def palindrome(x):
    return x == x[::-1]

pali = filter(palindrome, words)

list(pali)
```

```python
# Map Function
# Helps apply a function to each item/object in the list
# Calculate use the map()

# Convert a list of temperature from Celcius to Farenheit
def cel_fah(degree):
    return (degree * 9/5) + 32

temp_c = [10, 25, 45, 90]

# Apply the map function to convert C to F
temp_f = map(cel_fah, temp_c)

list(temp_f)
```

```python
# Convert String numbers to numbers
text = ["2", "4", "8", "10", "45"]

def text_num(x):
    num = int(x)
    return num

num_change = map(text_num, text)

list(num_change)
```

```python
# Lambda Function
# Also known as Anonymous Function
# Create quick function without defining it

# 18-30 age range
age = list(range(18,31))

# Add 10 years to people's age
age_10 = map(lambda x: x+10, age)

print(age)
list(age_10)
```

```python
# square each age
square_age = map(lambda x: x**2, age)

print(age)
list(square_age)
```

```python
# filter above 25
above_25 = filter(lambda x: x>25, age)

print(age)
list(above_25)
```
