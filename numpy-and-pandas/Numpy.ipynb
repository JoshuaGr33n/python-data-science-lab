## Access Numpy arrays

1. np.array[vertical index, horizontal index] - for one element (one dimension)
2. np.array[vertical start:vertical end, horizontal start:horizontal end] - for more than one elemnet (multi dimension)

```python
# Slicing 
grid_values
```

```python
# top left element
grid_values[0,0]
```

```python
# first row
grid_values[0,:]
```

```python
# first column
grid_values[:,0]
```

```python
#first two rows and first two columns
grid_values[0:2, 0:2]
```

#### Boolean array indexing

```python
# select all values greater than 2
# print(grid_values > 2)
grid_values[grid_values > 2]
```

```python
# select values less than 3 and greater than 5
grid_values[(grid_values < 3) | (grid_values > 5)]
```

#### Modifying Numpy Arrays

```python
# adding elements to an array
grid_values
```

```python
# axis = 0 append to rows (append a new row)
grid_values_extra =  np.append(grid_values, [[92, 88, 70]], axis=0)
grid_values_extra
```

```python
# axis = 1 append to columns (append a new column)
np_column = np.array([[92], [88], [70]])
grid_values_extra_extra =  np.append(grid_values_extra, np_column, axis=1)
grid_values_extra_extra
```

```python
# Removing elements
grid_values_del_row = np.delete(grid_values, [1], axis=0)
grid_values_del_row
```

#### Methods (arange, reshape, linspace, eye)

```python
# arange. return a range of numbers from 0-10
arr = np.arange(0, 10)
arr
```

```python
# return a list of numbers by 2 between 0 and 10
arr_step = np.arange(0,10,2)
arr_step
```

```python
# reshape. change the shape of the dimension from one to two
# arr_reshape = arr.reshape(2,5)
arr_reshape = arr.reshape(5,2)
arr_reshape
```

```python
# linspace. allows you to generate an array but make sure its evenly spaced
lin_space = np.linspace(0,20)
lin_space # each element add 0.408 to it till the end. they are evenly spaced

lin_space = np.linspace(0,20, 10) # to be spaced by 10
lin_space

lin_space = np.linspace(0,20, 5) # to be spaced by 5
lin_space

lin_space = np.linspace(0,20, 8) # to be spaced by 8
lin_space
```

```python
# eye. it gives a pattern among the elements
identity_matrix = np.eye(10)
identity_matrix
```

### Random: rand, randn, randint

```python
# rand: 0 & 1
rand_arr = np.random.rand(3,3)
rand_arr
```

```python
# randn: from a standard normal distribution anywhere between - & + with the mean 0 in between
randn_arr = np.random.randn(3,3)
randn_arr # some elements are negative while some are positive
```

```python
# randint: 1 & 100
randint_arr = np.random.randint(1,100, size=(3,3))
randint_arr
```
